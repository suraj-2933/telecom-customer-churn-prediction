```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,r2_score,f1_score
from sklearn.preprocessing import StandardScaler
```

```python
x = df_encode.drop(columns='Customer Status')
y = df_encode['Customer Status']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2 , random_state=42 ,    stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scale = scaler.fit_transform(x_train)
x_test_scale = scaler.transform(x_test)
```

```python
model_lr = LogisticRegression()
model_lr.fit(x_train_scale,y_train)
y_pred_lr = model_lr.predict(x_test_scale)
print(accuracy_score(y_test,y_pred_lr))
print(classification_report(y_test,y_pred_lr))
```

```python
cm_lr = confusion_matrix(y_test,y_pred_lr)
sns.heatmap(cm_lr ,annot=True,fmt='d')
plt.show()
```

```python
model_rf = RandomForestClassifier()
model_rf.fit(x_train,y_train)
y_pred_rf = model_rf.predict(x_test)
print(accuracy_score(y_test,y_pred_rf))
print(classification_report(y_test,y_pred_rf))
```

```python
cm_rf = confusion_matrix(y_test,y_pred_rf)
sns.heatmap(cm_rf ,annot=True,fmt='d')
plt.show()
```

```python
y_prob_rf = model_rf.predict_proba(x_test)[:, 1]

for threshold in [0.5, 0.45, 0.4, 0.35, 0.3]:
    y_pred = (y_prob_rf >= threshold).astype(int)

    print(f"\nThreshold: {threshold}")
    print(classification_report(y_test, y_pred, zero_division=0))
```

```python
from sklearn.metrics import roc_auc_score

y_prob_rf = model_rf.predict_proba(x_test)[:, 1]

print("Random Forest ROC-AUC:",
      roc_auc_score(y_test, y_prob_rf))
```

```python
from sklearn.metrics import RocCurveDisplay
RocCurveDisplay.from_predictions(
    y_test, y_prob_rf
)
plt.show()
```

### CROSS VALIDATION (CV)

```python
from sklearn.model_selection import cross_val_score
```

```python
scores_lr = cross_val_score(model_lr,x_train_scale,y_train,cv=5)
print(scores_lr)
print(scores_lr.mean())
```

```python
scores_rf = cross_val_score(model_rf,x_train,y_train,cv=5)
print(scores_rf)
print(scores_rf.mean())
```

### HYPERPARAMETER TUNING

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
param_grid_rf = {
    'n_estimators': [50,100,200],
    'max_depth': [None,10,20,30],
    'min_samples_split': [2,5,10],
    'min_samples_leaf': [1,2,4]
}
```

```python
random_search_rf = RandomizedSearchCV(estimator=model_rf,param_distributions=param_grid_rf,cv=5,
                                      n_iter=10 , random_state=42)
```

```python
random_search_rf.fit(x_train,y_train)
```

```python
print(random_search_rf.best_params_)
print(random_search_rf.best_score_)
```

```python
best_rf = random_search_rf.best_estimator_
```

```python
# Best practice: Run cross-validation on your tuned best estimator
scores_best_rf = cross_val_score(best_rf, x_train, y_train, cv=5)
print("Tuned RF Cross-Validation Scores:", scores_best_rf)
print("Tuned RF Mean Accuracy:", scores_best_rf.mean())
```

```python
# Get feature names directly from your encoded DataFrame columns (x)
feature_names = x.columns

# Get feature importances directly from the trained Random Forest model
importances = best_rf.feature_importances_

# Create the importance dataframe
importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importances
}).sort_values("Importance", ascending=False)

# Print the top 15 most important features
print(importance_df.head(15))
```
